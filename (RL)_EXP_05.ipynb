```python
# Experiment 5: Value Iteration
# Grid World

ROWS = 4
COLS = 4

# Goal and terminal states
GOAL = (0, 3)
TERMINAL = (3, 3)

# Actions
actions = {
    "UP": (-1, 0),
    "DOWN": (1, 0),
    "LEFT": (0, -1),
    "RIGHT": (0, 1)
}

# Parameters
gamma = 0.9
theta = 0.001


# Function to get next state
def get_next_state(state, action):

    row, col = state
    dr, dc = actions[action]

    new_row = row + dr
    new_col = col + dc

    # Stay in same state if outside grid
    if new_row < 0 or new_row >= ROWS:
        return state

    if new_col < 0 or new_col >= COLS:
        return state

    return (new_row, new_col)


# Function to get reward
def get_reward(next_state):

    if next_state == GOAL:
        return 10

    if next_state == TERMINAL:
        return -10

    return -1


# Initialize values
V = {}

for row in range(ROWS):
    for col in range(COLS):
        V[(row, col)] = 0


# Value Iteration
while True:

    delta = 0

    for row in range(ROWS):
        for col in range(COLS):

            state = (row, col)

            # Skip terminal states
            if state == GOAL or state == TERMINAL:
                continue

            old_value = V[state]

            action_values = []

            # Calculate value for every action
            for action in actions:

                next_state = get_next_state(state, action)

                reward = get_reward(next_state)

                value = reward + gamma * V[next_state]

                action_values.append(value)

            # Select maximum value
            V[state] = max(action_values)

            delta = max(delta, abs(old_value - V[state]))

    # Stop when values become stable
    if delta < theta:
        break


# Find optimal policy
policy = {}

for row in range(ROWS):
    for col in range(COLS):

        state = (row, col)

        if state == GOAL:
            policy[state] = "G"

        elif state == TERMINAL:
            policy[state] = "X"

        else:

            best_action = None
            best_value = float("-inf")

            for action in actions:

                next_state = get_next_state(state, action)

                reward = get_reward(next_state)

                value = reward + gamma * V[next_state]

                if value > best_value:

                    best_value = value
                    best_action = action

            # Convert action to arrow
            if best_action == "UP":
                policy[state] = "↑"

            elif best_action == "DOWN":
                policy[state] = "↓"

            elif best_action == "LEFT":
                policy[state] = "←"

            else:
                policy[state] = "→"


# Print Value Function
print("VALUE FUNCTION")
print("----------------")

for row in range(ROWS):

    for col in range(COLS):

        print(f"{V[(row, col)]:6.2f}", end=" ")

    print()


# Print Optimal Policy
print("\nOPTIMAL POLICY")
print("----------------")

for row in range(ROWS):

    for col in range(COLS):

        print(f"{policy[(row, col)]:^6}", end=" ")

    print()
```
